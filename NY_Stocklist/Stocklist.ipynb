```python
# import dependencies
import os
from selenium import webdriver
from selenium.webdriver.common.by import By
```

```python
# NY stock lists website 
# https://www.nyse.com/listings_directory/stock/
# Test URL: https://www.seleniumeasy.com/lander
# Set up driver
driver = webdriver.Chrome()
driver.get('https://www.nyse.com/listings_directory/stock/')

#implicit wait
driver.implicitly_wait(10)

# Find Table by class
table = driver.find_element(By.XPATH ,"//table[@class='table-data w-full table-border-rows']"
)

# Collect the rows
rows = table.find_elements(By.XPATH , './/tr')

# Define Dictionary to collect stock symbol and name
stocks = {'Symbol': [] , 'Name': []}

cells = rows[1].find_elements(By .XPATH , './/td')

stocks['Symbol'].append(cells[0].text)
stocks['Name'].append(cells[1].text)


# Iterate through each row and collect the Cells
# Then append each cell's value into their respective lists

# for row in rows[1:]:

#     cells = row.find_elements(By.XPATH , './/td')

#     for cell in range(0 , len(cells)):

#         if cell == 0:

#             stocks['Symbol'].append( cell)

#         else:

#             stocks['Name'].append( cell )









```
